```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import make_pipeline

# Generate synthetic data
np.random.seed(0)
X = np.sort(np.random.rand(30))  # 30 points between 0 and 1
y = np.sin(2 * np.pi * X) + np.random.normal(0, 0.1, X.shape)

X = X[:, np.newaxis]
y = y[:, np.newaxis]

# Define model degrees for underfitting, just right, and overfitting
degrees = [1, 3, 15]
titles = ['Underfitting', 'Just right!', 'Overfitting']

# Plotting
plt.figure(figsize=(15, 4))
for i, degree in enumerate(degrees):
    model = make_pipeline(PolynomialFeatures(degree), LinearRegression())
    model.fit(X, y)
    
    # Create test inputs for a smooth curve
    X_test = np.linspace(0, 1, 100)[:, np.newaxis]
    y_pred = model.predict(X_test)
    
    # Plot
    plt.subplot(1, 3, i + 1)
    plt.scatter(X, y, color='orange', label='Data')
    plt.plot(X_test, y_pred, color='blue', linewidth=2, label='Model')
    plt.title(titles[i])
    plt.xlabel('X')
    plt.ylabel('Y')
    plt.grid(True)

plt.tight_layout()

# Save the figure
output_path = "underfitting_vs_overfitting.png"
plt.savefig(output_path)
plt.close()

output_path

```
